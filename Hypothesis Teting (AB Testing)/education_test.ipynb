### A/B testing in python for student scores
---
- ## Step 1 - Define the Goal
# Does using tablets improve students test scores

```python
import numpy as np
from scipy.stats import ttest_ind

#generating a random seed for constant results
np.random.seed(23432)

#generation test scores
w_tablet = np.random.randint(40,60,50)
w_o_tablet = np.random.randint(25,60,50)

#perform t test
test_stat, p_value = ttest_ind(w_tablet,w_o_tablet)


#interprete results
alpha =  0.05
if p_value <= alpha:
    conclusion = "Reject H₀: Using Tablets increase students performance."
else:
    conclusion = "Fail to reject H₀: Using Tablets does not increase students performance"

print(conclusion)
```
