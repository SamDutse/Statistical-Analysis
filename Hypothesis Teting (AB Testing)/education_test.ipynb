# 🎓 Student Performance A/B Test

### Comparing Mathematics Scores: With vs Without Tablets

---

## 1️⃣ Define Goal

**Question:**  
Does using tablets in secondary schools improve students' mathematics scores?

I aim to evaluate whether integrating tablets into classroom learning has a **positive effect** on students' test performance.

---

## 2️⃣ Formulate Hypotheses

### ✅ Null Hypothesis (H₀):
> Using tablets **does not** improve students' math scores.  

### ✅ Alternative Hypothesis (H₁):
> Using tablets **does** improve students' math scores.  

This sets up a **one-tailed** test, focusing on whether tablets lead to **higher** scores.

---

## 3️⃣ Simulate Data and Store Scores

I simulated math test scores using `np.random.randint()`:

- **With Tablets (`w_tablet`)**:  
  Random scores between **40 and 60**, simulating better-prepared students.

- **Without Tablets (`w_o_tablet`)**:  
  Random scores between **25 and 60**, simulating more diverse performance with possibly lower averages.

I used `np.random.seed(2342)` to ensure **reproducibility**.

---

## 4️⃣ Perform Statistical Test

I used `scipy.stats.ttest_ind()` to run a **two-sample independent t-test** (Welch’s t-test, `equal_var=False`) to determine if the **mean scores** between groups differ significantly.

Significance level:  
\[
\alpha = 0.05
\]

---

## 5️⃣ Display Results & Interpret

I extracted:

- **Test Statistic (`t_stat`)** – indicates how far apart the group means are, in units of standard error.
- **P-Value** – the probability of observing such a difference if the null hypothesis were true.



```python
import numpy as np
from scipy.stats import ttest_ind
import matplotlib.pyplot as plt


#generating a random seed for constant result
np.random.seed(2342)

#generation test scores
w_tablet = np.random.randint(40,60,50)
w_o_tablet = np.random.randint(25,60,50)

#perform t test
test_stat, p_value = ttest_ind(w_tablet,w_o_tablet)


#interprete results
alpha =  0.05
if p_value <= alpha:
    conclusion = "Reject H₀: Using Tablets increase students performance."
else:
    conclusion = "Fail to reject H₀: Using Tablets does not increase students performance"

#interprete the result
print("Test Statistic:", test_stat)
print("P-Value:", p_value)
print("Conclusion:", conclusion)
```
